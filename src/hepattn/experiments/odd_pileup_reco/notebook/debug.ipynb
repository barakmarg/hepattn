```python
from pathlib import Path
from hepattn.experiments.odd_pileup_reco.eval_data import run_forward_pass, load_eval_data_from_h5
from hepattn.experiments.odd_pileup_reco.reco_analysis import run_reco_analysis, pflow_data_from_eval_dicts

CKPT_PATH = "/storage/agrp/barakma/hepattn/src/hepattn/experiments/odd_pileup_reco/logs/odd_pflow_reco_20260412-T163313/ckpts/epoch=028-val_loss=10.78560.ckpt"
CONFIG_PATH = "/storage/agrp/barakma/hepattn/src/hepattn/experiments/odd_pileup_reco/configs/base.yaml"
N_EVENTS = 1000
BATCH_SIZE = 64
BASE_TAG = "apr15_debug_true_mask_compare"

baseline_suff = f"{BASE_TAG}_predmask_n{N_EVENTS}"
debug_suff = f"{BASE_TAG}_truthmask_n{N_EVENTS}"

print("Running baseline forward pass...")
h5_baseline = run_forward_pass(
    ckpt_path=CKPT_PATH,
    config_path=CONFIG_PATH,
    files=val_files,
    num_events=N_EVENTS,
    batch_size=BATCH_SIZE,
    test_suff=baseline_suff,
    inference_mode=True,
    reco_debug_use_truth_masks=False,
    )

print("Running debug forward pass with truth masks in reco filtering...")
h5_debug = run_forward_pass(
    ckpt_path=CKPT_PATH,
    config_path=CONFIG_PATH,
    files=val_files,
    num_events=N_EVENTS,
    batch_size=BATCH_SIZE,
    test_suff=debug_suff,
    inference_mode=True,
    reco_debug_use_truth_masks=True,
    )

print("Baseline H5:", h5_baseline)
print("Debug H5:   ", h5_debug)
print("Baseline file name:", Path(h5_baseline).name)
print("Debug file name:   ", Path(h5_debug).name)

_, _, reco_baseline = load_eval_data_from_h5(h5_baseline, event_start=0, event_stop=N_EVENTS)
_, _, reco_debug = load_eval_data_from_h5(h5_debug, event_start=0, event_stop=N_EVENTS)

data_baseline = pflow_data_from_eval_dicts(reco_baseline)
data_debug = pflow_data_from_eval_dicts(reco_debug)

```
