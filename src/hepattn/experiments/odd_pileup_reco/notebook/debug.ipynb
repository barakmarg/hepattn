```python
import numpy as np
import torch
threshold = 1e-5    
event_index = 0
e = event_index
from hepattn.experiments.odd_pileup_reco.reco_analysis import select_charged_tracks
truth_full = np.asarray(reco_data["truth_incidence"][e], dtype=np.float32)   # (P, 5500)
pred_filt = np.asarray(reco_data["pred_incidence"][e], dtype=np.float32)     # (P, 1400)
reco_idx = np.asarray(reco_data["reco_node_indices"][e], dtype=np.int64)     # (1400,)
track_mask = np.asarray(reco_data["reco_is_track"][e]).astype(bool).reshape(-1)

# Map truth to filtered reco-node space
truth_filt = np.take(truth_full, reco_idx, axis=1)                             # (P, 1400)

# Raw row-0 diagnostics before enforcing constraints
row0_truth_track = truth_filt[0, track_mask]
row0_pred_track = pred_filt[0, track_mask]

print("=" * 72)
print("RAW DIAGNOSTICS")
print("=" * 72)
print("event:", e)
print("track columns:", int(track_mask.sum()))
print("truth row0 > 0 on track cols:", int(np.count_nonzero(row0_truth_track > 0)))
print("pred  row0 > 0 on track cols:", int(np.count_nonzero(row0_pred_track > 0)))
print("truth row0 max on track cols:", float(row0_truth_track.max()) if row0_truth_track.size else 0.0)
print("pred  row0 max on track cols:", float(row0_pred_track.max()) if row0_pred_track.size else 0.0)

# Enforce intended semantics for comparison:
# 1) threshold tiny values
# 2) force row0(track cols)=0
# 3) one-track-per-particle via select_charged_tracks
pred_proc = pred_filt.copy()
truth_proc = truth_filt.copy()

pred_proc[pred_proc <= threshold] = 0.0
truth_proc[truth_proc <= threshold] = 0.0

pred_proc[0, track_mask] = 0.0
truth_proc[0, track_mask] = 0.0

pred_sel = select_charged_tracks(
    torch.from_numpy(pred_proc).unsqueeze(0),
    torch.from_numpy(track_mask).unsqueeze(0),
).squeeze(0).cpu().numpy() > 0

truth_sel = select_charged_tracks(
    torch.from_numpy(truth_proc).unsqueeze(0),
    torch.from_numpy(track_mask).unsqueeze(0),
).squeeze(0).cpu().numpy() > 0

pred_tracks_per_particle = pred_sel[:, track_mask].sum(axis=1)
truth_tracks_per_particle = truth_sel[:, track_mask].sum(axis=1)

print("\n" + "=" * 72)
print("AFTER ENFORCING ROW0-TRACK=0 + ONE-TRACK SELECTION")
print("=" * 72)
print("pred max tracks/particle :", int(pred_tracks_per_particle.max()) if pred_tracks_per_particle.size else 0)
print("truth max tracks/particle:", int(truth_tracks_per_particle.max()) if truth_tracks_per_particle.size else 0)
print("pred particles with >1 track :", int(np.sum(pred_tracks_per_particle > 1)))
print("truth particles with >1 track:", int(np.sum(truth_tracks_per_particle > 1)))
print("row0 pred selected tracks  :", int(pred_tracks_per_particle[0]))
print("row0 truth selected tracks :", int(truth_tracks_per_particle[0]))

# Optional binary match stats excluding pileup row
pred_bin = pred_sel[1:, track_mask]
truth_bin = truth_sel[1:, track_mask]

matches = pred_bin & truth_bin
fp = pred_bin & (~truth_bin)
fn = truth_bin & (~pred_bin)

n_match = int(np.count_nonzero(matches))
n_fp = int(np.count_nonzero(fp))
n_fn = int(np.count_nonzero(fn))
n_truth = int(np.count_nonzero(truth_bin))
n_pred = int(np.count_nonzero(pred_bin))

recall = (100.0 * n_match / n_truth) if n_truth > 0 else 0.0
precision = (100.0 * n_match / n_pred) if n_pred > 0 else 0.0

print("\n" + "=" * 72)
print("MATCH STATS (EXCLUDING PILEUP ROW)")
print("=" * 72)
print("matches:", n_match, "fp:", n_fp, "fn:", n_fn)
print(f"recall={recall:.2f}%  precision={precision:.2f}%")
```

```python
import torch
import numpy as np

scaler = loader.dataset.scaler.transforms["pt"]

def sample_pred_pt_unique(n_batches=2, use_autocast=True):
    vals = []
    for bi, (inp, tgt) in enumerate(loader):
        if bi >= n_batches:
            break
        inp = {k: v.to(next(model.parameters()).device) if isinstance(v, torch.Tensor) else v for k, v in inp.items()}
        tgt = {k: v.to(next(model.parameters()).device) if isinstance(v, torch.Tensor) else v for k, v in tgt.items()}

        if use_autocast:
            ctx = torch.autocast(device_type="cuda", dtype=torch.bfloat16)
        else:
            ctx = torch.autocast(device_type="cuda", dtype=torch.bfloat16, enabled=False)

        with torch.no_grad(), ctx:
            out = model.model(inp, targets=tgt)
            pred = model.model.predict(out)

        pt = pred["reco_final"]["regression"]["reco_pflow_pt"].detach().cpu().float()
        tv = tgt["reco_particle_valid"].detach().cpu().bool()
        pt_inv = scaler.inverse_transform(pt.unsqueeze(-1)).squeeze(-1)
        vals.append(pt_inv[tv].numpy())

    vals = np.concatenate(vals)
    return {
        "n": int(vals.size),
        "uniq_0p01": int(np.unique(np.round(vals, 2)).shape[0]),
        "p1": float(np.percentile(vals, 1)),
        "p50": float(np.percentile(vals, 50)),
        "p99": float(np.percentile(vals, 99)),
        "max": float(np.max(vals)),
    }

print("autocast bf16:", sample_pred_pt_unique(use_autocast=True))
print("float32:     ", sample_pred_pt_unique(use_autocast=False))
```
