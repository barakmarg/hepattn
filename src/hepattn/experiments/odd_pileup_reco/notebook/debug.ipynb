```python
import torch
import numpy as np

dev = next(model.parameters()).device
inputs_b, targets_b = next(iter(loader))
inputs_b = {k: v.to(dev) if isinstance(v, torch.Tensor) else v for k, v in inputs_b.items()}
targets_b = {k: v.to(dev) if isinstance(v, torch.Tensor) else v for k, v in targets_b.items()}

# A) Current eval path: predict without loss (no Hungarian permutation)
with torch.no_grad(), torch.autocast(device_type=str(dev).split(":")[0], dtype=torch.bfloat16):
    out_a = model.model(inputs_b, targets=targets_b)
    pred_a = model.model.predict(out_a)

# B) Validation path: run loss first (permutes outputs in-place), then predict
with torch.no_grad(), torch.autocast(device_type=str(dev).split(":")[0], dtype=torch.bfloat16):
    out_b = model.model(inputs_b, targets=targets_b)
    _ = model.model.loss(out_b, targets_b)
    pred_b = model.model.predict(out_b)

pt_a = pred_a["reco_final"]["regression"]["reco_pflow_pt"].detach().cpu().float()
pt_b = pred_b["reco_final"]["regression"]["reco_pflow_pt"].detach().cpu().float()
truth_valid = targets_b["reco_particle_valid"].detach().cpu().bool()

scaler = loader.dataset.scaler.transforms["pt"]
pt_a_inv = scaler.inverse_transform(pt_a.unsqueeze(-1)).squeeze(-1)
pt_b_inv = scaler.inverse_transform(pt_b.unsqueeze(-1)).squeeze(-1)

vals_a = pt_a_inv[truth_valid].numpy()
vals_b = pt_b_inv[truth_valid].numpy()

print("No-loss->predict unique rounded (0.01):", np.unique(np.round(vals_a, 2)).shape[0])
print("Loss->predict unique rounded (0.01):   ", np.unique(np.round(vals_b, 2)).shape[0])
print("No-loss->predict min/max:", float(vals_a.min()), float(vals_a.max()))
print("Loss->predict min/max:   ", float(vals_b.min()), float(vals_b.max()))
```

```python
import torch
import numpy as np

scaler = loader.dataset.scaler.transforms["pt"]

def sample_pred_pt_unique(n_batches=2, use_autocast=True):
    vals = []
    for bi, (inp, tgt) in enumerate(loader):
        if bi >= n_batches:
            break
        inp = {k: v.to(next(model.parameters()).device) if isinstance(v, torch.Tensor) else v for k, v in inp.items()}
        tgt = {k: v.to(next(model.parameters()).device) if isinstance(v, torch.Tensor) else v for k, v in tgt.items()}

        if use_autocast:
            ctx = torch.autocast(device_type="cuda", dtype=torch.bfloat16)
        else:
            ctx = torch.autocast(device_type="cuda", dtype=torch.bfloat16, enabled=False)

        with torch.no_grad(), ctx:
            out = model.model(inp, targets=tgt)
            pred = model.model.predict(out)

        pt = pred["reco_final"]["regression"]["reco_pflow_pt"].detach().cpu().float()
        tv = tgt["reco_particle_valid"].detach().cpu().bool()
        pt_inv = scaler.inverse_transform(pt.unsqueeze(-1)).squeeze(-1)
        vals.append(pt_inv[tv].numpy())

    vals = np.concatenate(vals)
    return {
        "n": int(vals.size),
        "uniq_0p01": int(np.unique(np.round(vals, 2)).shape[0]),
        "p1": float(np.percentile(vals, 1)),
        "p50": float(np.percentile(vals, 50)),
        "p99": float(np.percentile(vals, 99)),
        "max": float(np.max(vals)),
    }

print("autocast bf16:", sample_pred_pt_unique(use_autocast=True))
print("float32:     ", sample_pred_pt_unique(use_autocast=False))
```
